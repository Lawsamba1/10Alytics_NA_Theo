Scenario: ShopSense is a small but growing phone accessories shop on Lagos Island. The store owner, Amara, tracks her products in notebook-but it's getting too slow and she keeps missingrestock warnings.

She has heard Python can help. She needs a simple script that:
  - Shows her which products are HOT, STEADY or Slow sellers.
  - Calculates total revenue and stock value.
  - Flags products that need restocking before she runs out.

```python
# --------------------------------------------
# 01. Define your Product Records
# --------------------------------------------
products = [
    {"name": "iPhone 15 Case", "price":2500.0, "units_sold":42, "stock_left":3, "is_featured":True, "category":"Cases"},
    {"name": "Screen Protector", "price":800.0, "units_sold":28,"stock_left":15, "is_featured":False, "category":"Protection"},
    {"name": "Wireless Charger", "price":3200.0, "units_sold":19,"stock_left":4, "is_featured":True, "category":"Charging"},
    {"name": "USB-C Hub", "price":5200.0, "units_sold":4,"stock_left":22, "is_featured":False, "category":"Accessories"},
    {"name": "AirPod Foam Tips", "price":650.0, "units_sold":35,"stock_left":8, "is_featured":False, "category":"Audio"},
     {"name": "Pop Socket", "price":450.0, "units_sold":12,"stock_left":30, "is_featured":True, "category":"Accessories"}
]
```

```python
# --------------------------------------------------------
# 02. Classify_Product()
# --------------------------------------------------------

def classify_product(product) :
    """Return HOT/STEADY/SLOW"""

    sold = product["units_sold"]

    if sold >=30:
        return "HOT"
    elif sold >=15:
        return "STEADY"
    else:
        return "SLOW"

print(classify_product(products[0]))





```

```python
# ------------------------------------------
# 03. Loop and Print the Shop Report 
# ------------------------------------------


separator = "==" * 48

print(separator)
print(f"{'SHOPSENSE NIGERIA - WEEKLY REPORT':^48}")
print(separator)

for p in products:
    status = classify_product(p)
    print(f" {p['name']:<22} ₦{p['price']:>7,.0f}  sold:{p['units_sold']:>3}  stock:{p['stock_left']:>3}  {status}")

print(separator)
```

```python

best_seller = products[0]
total_stock_value = 0

for p in products:

    # classification
    status = classify_product(p)

    # Best seller check
    if p["units_sold"] > best_seller["units_sold"]:
        best_seller = p

    # stock value accumulation
    total_stock_value += p["price"] * p["stock_left"]

   

```

```python
# 1 - Best seller
# Find the product with the highest units_sold. Loop through all products, track the one with the highest number-best seller, 
# print its name and how many it sold. Also print total stock value and restock alerts.
print("\n-------------------------------------------")
print("KEY BUSINESS INSIGHTS")
print("--------------------------------------------")

# Best seller
print(f"1. Best Seller: {best_seller['name']} ({best_seller['units_sold']} units sold.)")

# Total stock value 
print(f"2. Total Stock Value: ₦{total_stock_value:}")

# Restock alerts
print(f"3.Restock Alert:")
for p in products:
    if p["stock_left"] < 5 :
        print(f"- {p['name']} is low on stock ({p['stock_left']} left!)")


print("\n===  END OF REPORT  ===\n")


```
